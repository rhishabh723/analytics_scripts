```python
import pandas as pd
import pandas_gbq
import plotly.express as px
import sys, requests
from pathlib import Path
from zipfile import ZipFile
import urllib
from PIL import Image
```

```python
#Change the date to the previous day
query = """
SELECT 
PLAYER_NAME,PLAYER_ID, 
COUNT(*) as G,
SUM(CASE WHEN WL='W' THEN 1 ELSE 0 END) as Wins,
SUM(CASE WHEN WL='L' THEN 1 ELSE 0 END) as Losses,
ROUND(SUM(CASE WHEN WL='W' THEN 1 ELSE 0 END)/COUNT(*),2) as Win_Percent,
ROUND(SUM(CASE WHEN WL='L' THEN 1 ELSE 0 END)/COUNT(*),2) as Losses_Percent,
FROM `perceptive-ivy-290216.nba_api_player_boxscore.player_box_96_onwards`
GROUP BY 1,2
HAVING COUNT(*)>1000
ORDER BY 6 DESC
"""
project_id = "perceptive-ivy-290216"
df_bq2 = pandas_gbq.read_gbq(query, project_id=project_id, dialect='standard')
```

```python
fig = px.scatter(
  df_bq2.head(50), 
  x="G", 
  y="Win_Percent",
  title="<b>Player Career Regular Season Win% </b>",
  template="plotly_white",
  size="Win_Percent",
  height=800, 
  width=1200,
  hover_data=["PLAYER_NAME"],
  color="G",
  color_continuous_scale=px.colors.sequential.Agsunset_r,
  text='PLAYER_NAME',
  )

fig.update_layout(
    title_x=0.5,
    xaxis_title="<b>Games Played</b>",
    yaxis_title="<b>Regular Season Win Percentage</b>",
    hoverlabel=dict(
        bgcolor="white",
        font_size=16,
        font_family="PT Sans Narrow"
    ),
    yaxis = dict(tickfont = dict(size=20),tickformat=".0%"),
    xaxis = dict(tickfont = dict(size=20)),
    font=dict(
        family="PT Sans Narrow",
        size=12,
        color="Black"
    ),
    title_font_family="PT Sans Narrow"
)
fig.update_layout(showlegend=False)
fig.update_traces(textposition='top center')

fig
```
